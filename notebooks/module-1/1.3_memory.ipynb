## No memory

```python
from langchain.agents import create_agent


agent = create_agent(
    "ollama:gemma4"
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## Memory

```python
from langgraph.checkpoint.memory import InMemorySaver  
# Checkpoints store information for each thread, so that the previously included messages in the same agent thread can be referenced 
# For more information on checkpoints, see: https://reference.langchain.com/python/langgraph/checkpoints 
# There are other packages available to try and a comparison between each
# InMemorySaver will lose all checkpoints after a restart, as the checkpoints are in RAM. InMemorySaver is only recommended for debugging and testing
# In production, something like Postgres should be used
# For long-term storage, use stores
# For a comparison between checkpoints and stores, see: https://docs.langchain.com/oss/python/langgraph/persistence 

agent = create_agent(
    "ollama:gemma4",
    checkpointer=InMemorySaver(),  
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
