```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.tools import tool
from typing import Dict, Any
from tavily import TavilyClient
from langchain_community.utilities import SQLDatabase

tavily_client = TavilyClient()

db = SQLDatabase.from_uri("sqlite:///resources/Chinook.db")


@tool
def web_search(query: str) -> Dict[str, Any]:

    """Search the web for information"""

    return tavily_client.search(query)

@tool
def sql_query(query: str) -> str:

    """Obtain information from the database using SQL queries"""

    try:
        return db.run(query)
    except Exception as e:
        return f"Error: {e}"
```

```python
from dataclasses import dataclass

@dataclass
class UserRole:
    user_role: str = "external"
```

```python
from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse
from typing import Callable

@wrap_model_call
def dynamic_tool_call(request: ModelRequest, 
handler: Callable[[ModelRequest], ModelResponse]) -> ModelResponse:

    """Dynamically call tools based on the runtime context"""

    user_role = request.runtime.context.user_role
    
    if user_role == "internal":
        pass # internal users get access to all tools
    else:
        tools = [web_search] # external users only get access to web search
        request = request.override(tools=tools) 

    return handler(request)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="ollama:gemma4",
    tools=[web_search, sql_query],
    middleware=[dynamic_tool_call],
    context_schema=UserRole
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="How many artists are in the database?")]},
    context={"user_role": "internal"}
)

print(response["messages"][-1].content)
```

```python
# All of the middleware options may seem confusing! So here's a helpful matrix:

'''
Wrap-style Middleware
Uses: ModelRequest
What: Adjust an agent's tools, prompt, model while running
Eg: Giving an agent access to tools if the user is internal

Node-style Middleware
Uses: State + Runtime
What: Adjust an agent's state while running
Eg: Trimming messages from long conversations

Tool Calls
Uses: ToolRuntime
What: Agent adjusts its own state or retrieves runtime context
Eg: Updating the state with the username when provided to the agent

'''
```
